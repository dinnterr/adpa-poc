```
# experiments/results_analysis.ipynb (концепт)
import matplotlib.pyplot as plt

# Псевдо-дані для ілюстрації структури графіка
time_axis = range(0, 180, 5)
baseline_latency = [5.2]*36          # фіксований batch → стала висока latency
adaptive_latency = [0.8]*12 + [1.5]*12 + [0.9]*12  # ADPA: миттєво для critical

plt.plot(time_axis, baseline_latency, label="Baseline (fixed batch)")
plt.plot(time_axis, adaptive_latency, label="Proposed (ADPA)")
plt.xlabel("Time, s")
plt.ylabel("Latency, s")
plt.legend()
plt.title("Latency критичних подій: Baseline vs Adaptive")
```
